### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja1 = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

dfs = [loja1, loja2, loja3, loja4]
nomes_lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
```

# 1. Análise do faturamento



```python
import matplotlib.pyplot as plt

faturamentos = [df['Preço'].sum() for df in dfs]

plt.figure(figsize=(10, 5))
plt.bar(nomes_lojas, faturamentos, color='#4C72B0')
plt.title('Faturamento Total por Loja')
plt.ylabel('Faturamento (R$)')
plt.show()
```

# 2. Vendas por Categoria

```python

todas_lojas = pd.concat(dfs)
vendas_categoria = todas_lojas['Categoria do Produto'].value_counts()

plt.figure(figsize=(10, 6))
vendas_categoria.plot(kind='bar', color='#55A868')
plt.title('Total de Vendas por Categoria (Todas as Lojas)')
plt.ylabel('Quantidade de Itens Vendidos')
plt.xticks(rotation=45)
plt.show()
```

# 3. Média de Avaliação das Lojas


```python
avaliacoes = [df['Avaliação da compra'].mean() for df in dfs]

plt.figure(figsize=(10, 5))
plt.bar(nomes_lojas, avaliacoes, color='#C44E52')
plt.axhline(sum(avaliacoes)/4, color='black', linestyle='--', label='Média Geral')
plt.title('Avaliação Média dos Clientes')
plt.ylabel('Nota (1-5)')
plt.ylim(0, 5)
plt.legend()
plt.show()
```

# 4. Produtos Mais e Menos Vendidos

```python
print("TOP 5 PRODUTOS MAIS VENDIDOS")
print(todas_lojas['Produto'].value_counts().head(5))

print("\n5 PRODUTOS MENOS VENDIDOS")
print(todas_lojas['Produto'].value_counts().tail(5))
```

# 5. Frete Médio por Loja

```python
fretes = [df['Frete'].mean() for df in dfs]

plt.figure(figsize=(10, 5))
plt.bar(nomes_lojas, fretes, color='#8172B2')
plt.title('Custo Médio de Frete por Loja')
plt.ylabel('Valor do Frete (R$)')
plt.show()
```

# 6. Graficos

```python
import matplotlib.pyplot as plt
import pandas as pd

faturamentos = [df['Preço'].sum() for df in dfs]

plt.figure(figsize=(10, 6))
plt.bar(nomes_lojas, faturamentos, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
plt.title('Faturamento Total por Loja', fontsize=14)
plt.ylabel('Faturamento (R$)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

df_total = pd.concat(dfs)
contagem_categorias = df_total['Categoria do Produto'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(contagem_categorias, labels=contagem_categorias.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.title('Distribuição de Vendas por Categoria (Geral)', fontsize=14)
plt.show()

avaliacoes = [df['Avaliação da compra'].mean() for df in dfs]
fretes = [df['Frete'].mean() for df in dfs]

plt.figure(figsize=(10, 6))
plt.scatter(fretes, avaliacoes, s=200, color='purple', alpha=0.6)

for i, txt in enumerate(nomes_lojas):
    plt.annotate(txt, (fretes[i], avaliacoes[i]), xytext=(5, 5), textcoords='offset points')

plt.title('Relação: Frete Médio vs Avaliação Média', fontsize=14)
plt.xlabel('Custo Médio de Frete (R$)', fontsize=12)
plt.ylabel('Nota de Avaliação (1-5)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

# 7. Relatório final

1. Introdução
O objetivo desta análise foi avaliar o desempenho das quatro unidades da rede Alura Store para identificar qual delas apresenta a menor eficiência global. Para fundamentar a decisão de venda do Senhor João, cruzamos dados financeiros, logísticos e de satisfação do cliente.

2. Desenvolvimento e Análise dos Dados
Com base no processamento dos arquivos loja_1.csv a loja_4.csv, observamos os seguintes pontos:

Faturamento: A Loja 1 lidera o faturamento (aprox. R$ 1,53M), enquanto a Loja 4 apresenta o desempenho mais baixo (aprox. R$ 1,38M), uma diferença de quase R$ 150 mil.

Satisfação (Avaliações): Curiosamente, a loja que mais fatura (Loja 1) tem a pior média de avaliação (3.97), indicando problemas de qualidade ou atendimento. As Lojas 2 e 3 são as mais equilibradas, com notas acima de 4.0.

Logística (Frete): A Loja 4 possui o frete médio mais barato (R$ 31,27), o que não foi suficiente para impulsionar seu faturamento. A Loja 1 tem o frete mais caro (R$ 34,69), o que pode estar gerando as avaliações baixas.

Produtos e Categorias: O mix de produtos é similar entre as lojas, com categorias como "Eletrônicos" e "Móveis" dominando o volume de vendas.

3. Conclusão e Recomendação Final
A unidade recomendada para venda é a Loja 4.

Justificativa: Embora a Loja 4 tenha uma operação logística aparentemente barata (menor frete), ela não consegue converter isso em volume de vendas ou faturamento competitivo. Em um cenário de otimização de recursos, manter a loja com o menor faturamento total é o maior risco para o fluxo de caixa do grupo.

Sugestão Estratégica: Após a venda da Loja 4, recomenda-se que o Senhor João utilize parte do capital para:

Otimizar a Loja 1: Investigar por que o frete é tão alto e por que os clientes estão insatisfeitos, já que esta é a loja que traz mais dinheiro para a empresa.

Padronizar as Lojas 2 e 3: Seguir o modelo de sucesso dessas unidades, que mantêm um equilíbrio saudável entre lucro e satisfação.
